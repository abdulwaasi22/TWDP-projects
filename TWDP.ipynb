```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temps = [29, 31, 30, 33, 32]
humidity = [70, 65, 68, 75, 70]

plt.figure(figsize=(8,4))
plt.plot(days, temps,
         color="orange", marker="o")
plt.plot(days, humidity,
         color="blue", marker="o")
plt.title("Daily high temperature and Humidity, Lagos")
label="orange", "blue"
plt.legend(["temps", "humidity"])
plt.xlabel("Day")
plt.ylabel("Temperature (C), Humidtiy")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Recreate your Pandas Series
data = {
    'Abuja': 25.893699,
    'Lagos': 27.488767,
    'Tallahassee': 20.419726
}
temp_series = pd.Series(data, name="temp_c")

# 2. Create the bar plot
plt.figure(figsize=(8, 5))
temp_series.plot(kind='bar', color=['#4CAF50', '#FF9800', '#2196F3'], edgecolor='black')

# 3. Add labels and title
plt.title('Average City Temperatures', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('City', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.xticks(rotation=0)  # Keeps city names horizontal and easy to read
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 4. Display the chart
plt.tight_layout()
plt.show()
```
